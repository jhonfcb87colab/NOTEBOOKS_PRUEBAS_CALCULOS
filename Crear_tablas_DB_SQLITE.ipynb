```python
def Funcion_ejecutar_DDL(query_ddl, conexion=conn):
    """
    Funcion encargada de ejecutar sentencias de tipo DDL usando executescript
    para permitir múltiples sentencias SQL en una sola llamada.
    """
    cursor = conexion.cursor()
    
    # CAMBIO AQUÍ: executescript en lugar de execute
    cursor.executescript(query_ddl)
    
    conexion.commit()
    return f"DDL ejecutado correctamente."
```
